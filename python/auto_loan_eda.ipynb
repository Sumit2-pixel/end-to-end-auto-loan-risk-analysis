# Exploratory Data Analysis (EDA)

### EDA 1: Total Outstanding Loan Portfolio

```python
total_outstanding_balance = auto_loan_securitisation_data['CurrentBalance'].sum()
```

```python
total_outstanding_balance
```

**Business Insight:**

The company currently has an outstanding loan portfolio of ₹317.79 million, representing its total credit exposure. This portfolio requires continuous monitoring to identify high-risk loans, estimate expected credit losses, and support informed risk management decisions.

## EDA 2: IFRS 9 Stage Distribution

```python
stage_count = auto_loan_securitisation_data['IFRS9_Stage'].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(10,6))
plt.bar(x = stage_count.index , height = stage_count.values)
plt.title('IFRS 9 Stage Distribution')
plt.xlabel('IFRS 9 Stage')
plt.ylabel('Percentage (%)')
plt.show()
```

### Insight

- The majority of loans belong to **IFRS 9 Stage 1**, indicating that most borrowers are currently performing well and have a relatively low probability of default.
- A smaller proportion of loans fall under **Stage 2 and Stage 3**, representing accounts with significantly higher credit risk that require enhanced monitoring and collection efforts.
- Overall, the portfolio appears healthy, although the Stage 2 and Stage 3 population should remain a key focus for risk management.

## EDA 3: Top 5 States by Outstanding Loan Exposure

```python
state_wise_money_at_risk = auto_loan_securitisation_data.groupby('State')['CurrentBalance'].sum().sort_values(ascending=False)
```

```python
top_5_state_money_at_risk = state_wise_money_at_risk.head().sort_values()
```

```python
top_5_state_money_at_risk
```

```python
plt.figure(figsize=(10,6))
plt.barh(y = top_5_state_money_at_risk.index,width = top_5_state_money_at_risk.values)
plt.title('Top 5 state money at risk')
plt.xlabel('Amount')
plt.ylabel('state')
```

### Insight

- Chhattisgarh has the highest outstanding loan exposure among all states, followed by Gujarat and Goa.
- Higher outstanding balances indicate greater financial exposure for the lender in these regions.
- These states should receive closer portfolio monitoring, as any increase in delinquency could have a significant impact on overall portfolio performance.

## EDA 4: Outstanding Loan Exposure by Vehicle Type

```python
outstanding_loan_by_vehicle_type = auto_loan_securitisation_data.groupby('VehicleType')['CurrentBalance'].sum().sort_values(ascending=True)
```

```python
outstanding_loan_by_vehicle_type
```

```python
plt.figure(figsize=(10,6))
plt.barh(y = outstanding_loan_by_vehicle_type.index,width = outstanding_loan_by_vehicle_type.values)
plt.title('Outstanding Loan Exposure by Vehicle Type')
plt.xlabel('Outstanding Balance')
plt.ylabel('Vehicle Type')
plt.tight_layout()
plt.show()
```

### Insight

- SUV loans contribute the largest share of the outstanding portfolio, followed by MUVs.
- Hatchback loans represent the smallest portfolio exposure.
- Since SUVs account for the highest exposure, maintaining their credit quality is important for controlling overall portfolio risk.

## EDA 5: Distribution of Current Loan-to-Value (LTV)

```python
auto_loan_securitisation_data['LTV_Current'].head
```

```python
auto_loan_securitisation_data['LTV_Current'].describe()
```

```python
plt.figure(figsize=(10,6))

plt.hist(
    auto_loan_securitisation_data['LTV_Current'],
    bins=20,
    edgecolor='black'
)

plt.title('Distribution of Current Loan-to-Value (LTV)')
plt.xlabel('Current LTV')
plt.ylabel('Number of Loans')

plt.tight_layout()
plt.show()
```

### Insight

- Most loans are concentrated within the moderate LTV range, indicating a balanced collateral position for the majority of borrowers.
- Only a small number of loans exhibit very high LTV values, which may represent higher credit risk due to lower borrower equity.
- Monitoring high-LTV loans can help reduce potential losses during default scenarios.
